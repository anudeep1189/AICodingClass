```python
import pandas as pd
import numpy as np
from sklearn.datasets import fetch_california_housing
# Load the dataset
california = fetch_california_housing()
```

```python
# q1
df = pd.DataFrame(data = california.data, columns=california.feature_names)

df['MedHouseVal'] = california.target
df.head()
df.columns
```

```python
#q8 
maxVal = df['MedHouseVal'].idxmax()
maxHouseRow = df.loc[maxVal]
print(maxHouseRow)
print(f"Population -> {maxHouseRow['Population']}")
```

```python
#q9
df.loc[:49, "AveBedrms"]=np.nan
print(df["AveBedrms"].isnull().sum())
```

```python
#q10
medianValue = df["AveBedrms"].median()
df["AveBedrms"] = df["AveBedrms"].fillna(medianValue)
print(df["AveBedrms"].isnull().sum())
```

```python
#q11
df = pd.concat([df,df.head(5)],ignore_index=True)
duplicatesRows = df.duplicated().sum()
print(f"Duplicate rows = {duplicatesRows}")

df = df.drop_duplicates(ignore_index=True)
print(f"After droping = {df.duplicated().sum()}")
```

```python
#12
df["HouseAge"] = df["HouseAge"].mask(df["HouseAge"] > 40 , 40.0 )

print(f"max house age = {df["HouseAge"].max()}")
```

```python
#q13
catagoryAge = [0,15,30,100]
Catagorylabels = ["New","Established","Hostoric"]
df["AgeCategory"] = pd.cut(df["HouseAge"],bins=catagoryAge,labels=Catagorylabels)

print(df["AgeCategory"].value_counts())
```

```python
#Q14
grouped_df = (
    df.groupby("AgeCategory", observed=False)[["MedHouseVal", "MedInc"]]
    .mean()
)

print(grouped_df)
```

```python
#q15
pivotTable = pd.pivot_table(
    df,
    values="Population",
    index="AgeCategory",
    aggfunc="max",
    observed=False,
)

print(pivotTable)
```

```python
#16
age_freq_table = pd.crosstab(
    index=df["AgeCategory"], 
    columns="Count", 
    colnames=[""]
)
print(age_freq_table) 
```

```python
#17
df["Rooms_per_Occupant"] = df["AveRooms"] / df["AveOccup"]
print(df[["AveRooms", "AveOccup", "Rooms_per_Occupant"]].head())
```

```python
#18
mean_inc = df["MedInc"].mean()
std_inc = df["MedInc"].std()
df["MedInc_Standardized"] = (df["MedInc"] - mean_inc) / std_inc
print("Standardized Mean:", round(df["MedInc_Standardized"].mean(), 4))
print("Standardized Std:", round(df["MedInc_Standardized"].std(), 4))
```

```python
#19
df["Is_Expensive"] = (df["MedHouseVal"] > 3.0).astype(int)
print(df["Is_Expensive"].value_counts())
```

```python
#20
df = df.sort_values(by="Longitude", ascending=True).reset_index(drop=True)
df["Spatial_Rolling_Price"] = (
    df["MedHouseVal"].rolling(window=5).mean()
)
print(df[["Longitude", "MedHouseVal", "Spatial_Rolling_Price"]].head(10))
```
