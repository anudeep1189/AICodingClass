```python
import pandas as pd 
from sklearn.datasets import fetch_california_housing
import numpy as np
import matplotlib.pyplot as plt
import random
```

```python
bins = [0,10,30,40,50,60]
```

```python
h_data = fetch_california_housing()
```

```python
df = pd.DataFrame(h_data.data,columns=h_data.feature_names)
df
```

```python
df[h_data.target_names[0]] = h_data.target
```

```python
id = random.sample(range(len(df)),1000)
df_random = df.iloc[id]
plt.scatter(df_random["HouseAge"], df_random["MedHouseVal"])
```

```python
plt.scatter(df_random["HouseAge"], df_random["MedHouseVal"],df_random["AveRooms"],cmap='virdis')
```

```python
sc = plt.scatter(df_random["HouseAge"], df_random["MedHouseVal"],c=df_random["AveRooms"],cmap='viridis',alpha=1)
plt.colorbar(sc,label="Average Rooms")
plt.title("House age vs house value")
plt.xlabel("House age")
plt.ylabel("Med House Value")
```
